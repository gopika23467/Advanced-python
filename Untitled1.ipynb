```python
numbers=["8592828136","76853","1234567890","4356536726","4356853682"]
invalid_count=0
valid_count=0
valid_nums=set()
for num in numbers:
  if len(num)!=10:
    print("invalid")
    invalid_count+=1
    continue
  else:
    print("valid")
    valid_count += 1
    valid_nums.add(num)


print(f"invalid_cound is {invalid_count}and valid is{valid_count}")
print(valid_nums)
```

```python
num=int(input("enter a number"))
if num <2:
  print("not a prime number")
else:
  for i in range(2,num):
    if num%i==0:
      is_prime =False
      break
if is_prime:
      print("prime number")
else:
      print("not a prime number")
```
